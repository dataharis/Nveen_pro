```python
import pandas as pd

df = pd.read_csv("power_grid_corporation\power_grid_corp_power_demand.csv")

df.head()
```

**Since we are working on a time series analysis. we have to sort with the respect to time**

```python
df['datetime'] = pd.to_datetime(df['datetime'], format='%Y-%m-%d %H:%M:%S')
df.set_index('datetime', inplace=True)

df.sort_index(inplace=True)
df.head()
```

```python
df.shape
```

```python
df.info()
```

**the values that are nan mean no data for this we can change it to 0**

```python
df['solar'].fillna(0, inplace=True)
df['wind'].fillna(0, inplace=True)
df['india_adani'].fillna(0, inplace=True)
df['nepal'].fillna(0, inplace=True)
print(df.info())
df.head()
```

```python
df.head()
```

```python
df['hour'] = df.index.hour # adding new feature on time based
df['dayofweek'] = df.index.dayofweek
df['month'] = df.index.month
df['year'] = df.index.year
df['dayofyear'] = df.index.dayofyear
df['is_weekend'] = (df.index.dayofweek >= 5).astype(int)

df.shape

```

```python
df.head()
```

```python
# Adding new features on lag bas hourly,weekly etc.
df['demand_lag_24h'] = df['demand_mw'].shift(24) 
df['demand_lag_168h'] = df['demand_mw'].shift(168)
df['demand_lag_1h'] = df['demand_mw'].shift(1)

df.shape # should be 22 columns
```
